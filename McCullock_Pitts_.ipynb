# Implementación de la Neurona de McCulloch-Pitts

## Neurona Compuerta OR

```python
# Definir la función de activación (Función Escalón)
def activation_function(z, threshold):
    return 1 if z >= threshold else 0

# Modelo de neurona McCulloch-Pitts para la compuerta OR
def neuron_or(x1, x2):
    # Definir pesos y umbral
    w1 = 1
    w2 = 1
    threshold = 1

    # Calcular la suma ponderada
    z = (x1 * w1) + (x2 * w2)

    # Salida final aplicando la función de activación
    return activation_function(z, threshold)

# Tabla de verdad para verificar la funcionalidad
inputs_or = [(0, 0), (0, 1), (1, 0), (1, 1)]

print("Tabla de verdad para la neurona OR:")
print("x1 | x2 | Resultado")
print("-------------------")
for x1, x2 in inputs_or:
    result = neuron_or(x1, x2)
    print(f" {x1} |  {x2} |    {result}")
```

### Explicación de la Neurona OR
En esta simulación, la neurona de McCulloch-Pitts actúa como una compuerta lógica OR. La neurona recibe dos entradas (x₁ y x₂), las cuales se multiplican por un peso positivo (w₁ = 1, w₂ = 1). Si al menos una de las entradas es 1, la suma ponderada alcanza o supera el umbral (threshold = 1), lo que activa la neurona y genera un 1 como salida. Permanece inactiva (salida 0) solo cuando ambas entradas son 0.

## Neurona Compuerta NOT

```python
import numpy as np

# Definir la función de activación (Función Escalón)
def activation_function(z, threshold):
    return 1 if z >= threshold else 0

# Modelo de neurona McCulloch-Pitts para la compuerta NOT
def neuron_not(x):
    # Definir peso inhibitorio y umbral
    w = -1
    threshold = 0

    # Calcular la suma ponderada
    z = x * w

    # Salida final aplicando la función de activación
    return activation_function(z, threshold)

# Tabla de verdad para verificar la funcionalidad
inputs_not = [0, 1]

print("Tabla de verdad para la neurona NOT:")
print("x | Resultado")
print("-------------")
for x in inputs_not:
    result = neuron_not(x)
    print(f"{x} |    {result}")
```
